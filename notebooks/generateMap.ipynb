```python
# =========================
# map creation
# =========================
from pathlib import Path
import folium
import pandas as pd
from selenium import webdriver
outdir = Path.cwd().parent / "results" / "heatmap"

m = folium.Map(
    location=[-36.8485, 174.7633], 
    zoom_start=13, 
    tiles='OpenStreetMap',
    max_bounds=True,
    zoom_snap=0,       # no snapping (true continuous zoom)
    zoom_delta=0.25,   
)

df = pd.read_csv("../data/at-dataset/SCATS-data/Site-List-Coordinates.csv")

for _, r in df.iterrows():
    
    folium.Marker(
        [r.Latitude, r.Longitude],
        icon=folium.DivIcon(
            icon_size=(48, 24),
            icon_anchor=(0, 0),
            html=f"<div style='height: 10px; width: 10px; background:rgba(255,0,0); border-radius:50%; display:inline-block'></div><div style='font-size:20px; font-weight:600; background:rgba(255,255,0,0.7); "
                f"padding:2px 4px; border-radius:4px;'>{int(r.Detector_ID)}</div>"
        )
    ).add_to(m)

bounds = [[-36.84217504552652, 174.75221065279757], [-36.86144802510803, 174.78761920956393]]

folium.TileLayer(
    location=[-36.8485, 174.7633],
    zoom_start=13,
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}",
    attr="Esri, Maxar, Earthstar Geographics",
    name="Esri WorldStreetMap",
    overlay=False,
    control=True,
).add_to(m)

# Optional: Draw rectangle outline for clarity
folium.Rectangle(bounds=bounds, color="blue", fill=False).add_to(m)

# Apply restriction
m.fit_bounds(bounds)

# folium.TileLayer("CartoDB dark_matter").add_to(m)
# folium.TileLayer("Stamen Terrain").add_to(m)
# folium.TileLayer("Stamen Toner").add_to(m)
# folium.TileLayer("Stamen Watercolor").add_to(n)
# folium.LayerControl().add_to(n)
m.save(outdir / "map.html")   # open in browser
```
